```python
#Analisis Hipotesis

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
import warnings
from statsmodels.tools.sm_exceptions import SingularMatrixWarning

# Sembunyikan peringatan yang tidak perlu agar output terminal bersih
warnings.simplefilter('ignore', SingularMatrixWarning)

df = pd.read_csv(sheet_url)

# -------------------------------------------------------------
# STEP 1 & 2: MUAT DATA DAN GANTI NAMA KOLOM
# -------------------------------------------------------------

rename_mapping = {
    df.columns[2]: 'Usia', df.columns[3]: 'Jenis_Kelamin', df.columns[4]: 'Pekerjaan',
    df.columns[5]: 'Punya_Rekening', df.columns[6]: 'Pernah_Pakai_AI', df.columns[7]: 'Profil_Risiko',
    df.columns[8]: 'AI_1', df.columns[9]: 'AI_2', df.columns[10]: 'AI_3', df.columns[11]: 'AI_4', df.columns[12]: 'AI_5',
    df.columns[13]: 'SE_1', df.columns[14]: 'SE_2', df.columns[15]: 'SE_3', df.columns[16]: 'SE_4', df.columns[17]: 'SE_5', df.columns[18]: 'SE_6',
    df.columns[19]: 'Minat_1', df.columns[20]: 'Minat_2', df.columns[21]: 'Minat_3',
    df.columns[22]: 'Lit_1', df.columns[23]: 'Lit_2', df.columns[24]: 'Lit_3',
    df.columns[25]: 'LPS_1', df.columns[26]: 'LPS_2', df.columns[27]: 'LPS_3'
}
df.rename(columns=rename_mapping, inplace=True)

# -------------------------------------------------------------
# STEP 3: KONVERSI SKALA LIKERT KE ANGKA
# -------------------------------------------------------------
likert_map = {'Sangat Tidak Setuju': 1, 'Tidak Setuju': 2, 'Netral': 3, 'Setuju': 4, 'Sangat Setuju': 5}
ai_cols = ['AI_1', 'AI_2', 'AI_3', 'AI_4', 'AI_5']
se_cols = ['SE_1', 'SE_2', 'SE_3', 'SE_4', 'SE_5', 'SE_6']
minat_cols = ['Minat_1', 'Minat_2', 'Minat_3']

# Terapkan mapping dan hapus baris yang kosong (NaN)
for col in ai_cols + se_cols + minat_cols:
    df[col] = df[col].astype(str).str.strip().map(likert_map)
df = df.dropna(subset=ai_cols + se_cols + minat_cols)

# -------------------------------------------------------------
# STEP 4: PERHITUNGAN SKOR LITERASI KEUANGAN (W)
# -------------------------------------------------------------
# Kunci jawaban kuis berdasarkan teori Lusardi & Mitchell
df['Lit_1_score'] = (df['Lit_1'].astype(str).str.contains('Lebih dari Rp1.100.000', case=False, na=False)).astype(int)
df['Lit_2_score'] = (df['Lit_2'].astype(str).str.contains('Lebih sedikit', case=False, na=False)).astype(int)
df['Lit_3_score'] = (df['Lit_3'].astype(str).str.contains('Menyebar modal investasi', case=False, na=False)).astype(int)

# Total skor W (0 hingga 3)
df['W'] = df['Lit_1_score'] + df['Lit_2_score'] + df['Lit_3_score']

# -------------------------------------------------------------
# STEP 5: PEMBUATAN VARIABEL KOMPOSIT DAN PENGELOMPOKAN KONTROL
# -------------------------------------------------------------
# Hitung rata-rata X, M, dan Y
df['X'] = df[ai_cols].mean(axis=1)
df['M'] = df[se_cols].mean(axis=1)
df['Y'] = df[minat_cols].mean(axis=1)

# Bersihkan dan siapkan variabel kontrol
df['Usia'] = pd.to_numeric(df['Usia'], errors='coerce')

# PENGELOMPOKAN PEKERJAAN (Modifikasi Baru)
kategori_utama = ['Karyawan Swasta', 'TNI/POLRI/PNS/BUMN', 'Profesional']
df['Pekerjaan'] = df['Pekerjaan'].apply(lambda x: x if x in kategori_utama else 'Lainnya')

# Ekstrak kata pertama untuk profil risiko (Konservatif, Moderat, Agresif)
df['Profil_Risiko'] = df['Profil_Risiko'].astype(str).apply(lambda x: x.split()[0] if pd.notnull(x) else 'Unknown')
df = df.dropna(subset=['Usia', 'Pekerjaan', 'Profil_Risiko'])

# Dummy coding untuk Pekerjaan dan Profil Risiko
df_controls = pd.get_dummies(df[['Pekerjaan', 'Profil_Risiko']], drop_first=True, dtype=float)
# Hapus spasi dari nama kolom dummy agar tidak error di statsmodels
df_controls.columns = [c.replace(' ', '_').replace('/', '_') for c in df_controls.columns]
df = pd.concat([df, df_controls], axis=1)
control_vars = " + ".join(df_controls.columns.tolist()) + " + Usia"

# -------------------------------------------------------------
# STEP 6: MEAN-CENTERING (Agar Interaksi Moderasi Stabil)
# -------------------------------------------------------------
df['M_c'] = df['M'] - df['M'].mean()
df['W_c'] = df['W'] - df['W'].mean()
df['MW_interact'] = df['M_c'] * df['W_c']  # Variabel moderasi M x W

# -------------------------------------------------------------
# STEP 7: UJI REGRESI OLS (H1, H2, H4)
# -------------------------------------------------------------
print("="*60)
print("HASIL UJI MODEL M (H1: X -> M)")
print("="*60)
formula_m = f"M ~ X + {control_vars}"
model_m = smf.ols(formula=formula_m, data=df).fit()
print(model_m.summary().tables[1])
# Interpretasi H1: Jika P>|t| untuk variabel X berada di bawah 0.05, maka H1 Diterima.

print("\n" + "="*60)
print("HASIL UJI MODEL Y (H2: M -> Y | H4: M*W -> Y)")
print("="*60)
formula_y = f"Y ~ X + M_c + W_c + MW_interact + {control_vars}"
model_y = smf.ols(formula=formula_y, data=df).fit()
print(model_y.summary().tables[1])
# Interpretasi H2: Jika P>|t| untuk M_c di bawah 0.05, maka H2 Diterima.
# Interpretasi H4: Jika P>|t| untuk MW_interact di bawah 0.05, maka H4 Diterima.

# -------------------------------------------------------------
# STEP 8: BOOTSTRAPPING UNTUK UJI MEDIASI (H3) & MODERATED-MEDIATION (H5)
# -------------------------------------------------------------
print("\n" + "="*60)
print("HASIL UJI BOOTSTRAPPING 5000 SAMPEL (H3 & H5)")
print("="*60)

np.random.seed(42)
n_bootstraps = 5000
boot_indirect_effects = []
boot_index_mod_med = []

# Path 'a' adalah koefisien dari X menuju M (Diambil dari model_m)
a_path = model_m.params['X']

for _ in range(n_bootstraps):
    # Resample dataset secara acak
    boot_samp = df.sample(n=len(df), replace=True)

    # Hitung ulang koefisien dengan data resample
    boot_model_m = smf.ols(formula=formula_m, data=boot_samp).fit()
    boot_model_y = smf.ols(formula=formula_y, data=boot_samp).fit()

    boot_a = boot_model_m.params['X']
    boot_b1 = boot_model_y.params['M_c']          # Jalur b1 (M -> Y)
    boot_b2 = boot_model_y.params['MW_interact']  # Jalur moderasi (Interaksi)

    # Efek Mediasi Langsung (pada rata-rata moderator karena sudah di-mean-center)
    boot_indirect_effects.append(boot_a * boot_b1)

    # Index of Moderated Mediation (a * b2)
    boot_index_mod_med.append(boot_a * boot_b2)

# Hitung Interval Kepercayaan 95% (Lower Level dan Upper Level)
ci_indirect = np.percentile(boot_indirect_effects, [2.5, 97.5])
ci_mod_med = np.percentile(boot_index_mod_med, [2.5, 97.5])

print(f"[H3] Efek Tidak Langsung (Mediasi):")
print(f"Estimasi Nilai: {np.mean(boot_indirect_effects):.4f}")
print(f"95% Confidence Interval: [{ci_indirect[0]:.4f} hingga {ci_indirect[1]:.4f}]")
if ci_indirect[0] > 0 or ci_indirect[1] < 0:
    print("-> Kesimpulan: H3 DITERIMA (Mediasi terbukti karena angka 0 TIDAK berada di dalam rentang interval)\n")
else:
    print("-> Kesimpulan: H3 DITOLAK (Tidak ada mediasi yang signifikan)\n")

print(f"[H5] Index of Moderated Mediation:")
print(f"Estimasi Nilai: {np.mean(boot_index_mod_med):.4f}")
print(f"95% Confidence Interval: [{ci_mod_med[0]:.4f} hingga {ci_mod_med[1]:.4f}]")
if ci_mod_med[0] > 0 or ci_mod_med[1] < 0:
    print("-> Kesimpulan: H5 DITERIMA (Literasi Keuangan secara signifikan memoderasi jalur mediasi AI -> Minat)")
else:
    print("-> Kesimpulan: H5 DITOLAK (Literasi Keuangan tidak mampu memoderasi mediasi tersebut)")
```
